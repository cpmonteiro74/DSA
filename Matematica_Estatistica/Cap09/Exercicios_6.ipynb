### Exercício 1: Soma de Matrizes
Escreva  um  programa  em  Python  que  soma  duas  matrizes  de mesmo  tamanho.
As matrizes devem ser lidas como listas de listas e o resultado deve ser exibido na tela.
Apresente a solução com e sem o uso do NumPy.
Pratique suas habilidades de programação ao mesmo tempo que estuda Matemática.

```python
import numpy as np
```

```python
a=[[1,2,3],\
   [4,5,6]]
b=[[7,8,9],\
   [0,1,2]]
```

```python
def cria_matriz(n_linhas, n_colunas, valor):
    matriz = [] # Lista Vazia para matriz
    for x in range(n_linhas):
        #Cria a linha x
        linha = [] # lista vazia para linhas
        for y in range(n_colunas):
            linha.append(valor)

        # Adiciona linha à matriz
        matriz.append(linha)
    
    return matriz    

def soma_matrizes(m1, m2):
    num_lin = len(m1)
    num_col = len(m1[0])

    num_linB = len(m2) # Verifica quantidade de linhas
    num_colB= len(m2[0]) # Verifica quantidade de colunas
    
    if num_lin == num_linB and num_col == num_colB:

        C = cria_matriz(num_lin, num_col, 0) # Chama A função cria_matriz

        for lin in range(num_lin): # Percorre linhas da matriz
            for col in range(num_col): # Percorre colunas da matriz
                C[lin][col] = m1[lin][col] + m2[lin][col]
        return C
    else:
        return False  
```

```python
soma_matrizes(a,b)
```

```python
np.array(a)+np.array(b)
```

### Exercício 2: Multiplicação de Matrizes
Implemente  uma  função  em  Python  que  multiplica  duas  matrizes.
Verifique  se  as dimensões  das  matrizes  são  compatíveis  para  a  multiplicação  e  retorne  a  matriz  resultante. 
Apresente a solução com e sem o uso do NumPy.
Pratique suas habilidades de programação ao mesmo tempo que estuda Matemática.

```python
def multiplica_matrizes(m1, m2):
    num_lin = len(m1)
    num_col = len(m1[0])

    num_linB = len(m2) # Verifica quantidade de linhas
    num_colB= len(m2[0]) # Verifica quantidade de colunas
    
    if num_lin == num_linB and num_col == num_colB:

        C = cria_matriz(num_lin, num_col, 0) # Chama A função cria_matriz

        for lin in range(num_lin): # Percorre linhas da matriz
            for col in range(num_col): # Percorre colunas da matriz
                C[lin][col] = m1[lin][col] * m2[lin][col]
        return C
    else:
        return False
```

### Exercício 3: Transposição de Matriz
Crie um programa que calcula a transposta de uma matriz.
A matriz deve ser inserida pelo usuário e o programa deve exibir a matriz transposta.
Apresente a solução com e sem o uso do NumPy.
Pratique suas habilidades de programação ao mesmo tempo que estuda Matemática.

```python
def transposta(m):
    num_lin = len(m)
    num_col = len(m[0])

    num_linT = num_col # linhas da transposta
    num_colT = num_lin # colunas da transposta
    
    T = cria_matriz(num_linT, num_colT, 0) # Chama A função cria_matriz

    for lin in range(num_linT): # Percorre linhas da matriz
        for col in range(num_colT): # Percorre colunas da matriz
                T[lin][col] = m[col][lin]
    return T    
```

```python
transposta(a)
```

```python
np.transpose(a)
```

### Exercício 4: Traço de uma Matriz
Desenvolva um script em Python que calcula o traço de uma matriz (soma dos elementos da  diagonal  principal).
O  programa  deve  funcionar  para  matrizes  quadradas  de  qualquer tamanho.
Apresente  a  solução  com  e  sem  o  uso  do  NumPy.
Pratique  suas  habilidades  de programação ao mesmo tempo que estuda Matemática.

```python
def soma_diagonal(m):
    num_lin = len(m)
    num_col = len(m[0])
    traco=0
      
    if num_lin == num_col: 
        for lin in range(num_lin): # Percorre linhas da matriz
            traco+= m[lin][lin]
        return traco
    else:
        return False 
```

```python
def cria_matriz(n_linhas, n_colunas, valor):
    matriz = [] # Lista Vazia para matriz
    for x in range(n_linhas):
        #Cria a linha x
        linha = [] # lista vazia para linhas
        for y in range(n_colunas):
            linha.append(valor)

        # Adiciona linha à matriz
        matriz.append(linha)
```

```python
m=[[1,2,3,4],\
   [1,1,4,9],\
   [1,2,1,4],\
   [1,3,4,1]]
```

```python
def matriz_cofator(m):
    matriz=[]
    for i in range(len(m)-1):
        linha=[]
        for i in range(len(m)-1):
            linha.append(0)
        matriz.append(linha)
    return matriz

def determinante(m):
    num_lin = len(m)
    num_col = len(m[0])
          
    if num_lin == num_col:

        C = matriz_cofator(m) # Chama A função cria_matriz

        for lin in range(num_lin-1): # Percorre linhas da matriz
            for col in range(num_col-1): # Percorre colunas da matriz
                C[lin][col] = m[lin-1][col-1]
        return C
    else:
        return False 

```
