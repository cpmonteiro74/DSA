## Exercício 1: Soma de Vetores
Descrição:  Escreva  uma  função  em  Python  que  some  dois  vetores.  Suponha  que  os 
vetores  têm  o  mesmo  tamanho.  Resolva  o  exercício  de  forma  diferente  do  que  foi 
mostrado nas 
aulas.

```python
V1=[1,4,3,5,6]
V2=[7,3,4,2,4]
```

```python
def soma_vetores(a,b):
    """
    Esta função retorna a soma entre dois vetores
    Os vetores a,b devem posuir mesmo tamanho
    """
    return [a+b for a,b in zip(a,b)]
```

```python
soma_vetores(V1,V2)
```

### Exercício 2: Produto Escalar

Descrição:  Implemente  uma  função  que  calcule  o  produto  escalar  entre  dois  vetores. 
Resolva o exercício de forma diferente do que foi mostrado nas aulas.

```python
def dot_product(a,b):
    """
    Esta função retorna o produto escalar entre dois vetores
    Os vetores a,b devem posuir mesmo tamanho
    """
    return sum([a[i] * b[i] for i in range(len(a))])
```

```python
dot_product(V1,V2)
```

### Exercício 3: Norma de um Vetor
Descrição: Escreva uma 
função que calcule a norma (ou comprimento) de um vetor.

```python
import numpy as np
def norma(a):
    """
    Esta função retorna a norma (magnitude) de um vetor
    """
    return np.sqrt(sum(i**2 for i in a))
```

```python
norma(V1)
```

### Exercício 4: Ângulo entre Dois Vetores
Descrição: Implemente uma função que calcule o ângulo (em graus) entre dois vetores.

```python
def angulo_vectors(a,b):

    """
    Calcula o angulo ente dois vetores de mesmas dimensões
        cos(aôb)=produto escalar(a,b)/ norma(a) * norma(b)
        angulo(aôb)=arccos(cos(aôb)
        angulo(aôb) em graus = angulo(aôb) * 180/2pi
    """
    
    #Calcula o produto interno entre os vetores
    dot_p=sum([a[i] * b[i] for i in range(len(a))])

    #calculo da norma de um vetor |x|
    def norma(x):
        return np.sqrt(sum([x[i]**2 for i in range(len(x))]))
    
    #Calcula o angulo em graus formado pelos vetores
    return np.degrees(np.arccos(dot_p/(norma(a)*norma(b))))
```

```python
angulo_vectors(V1,V2)
```

### Exercício 5: Projeção de um Vetor sobre Outro
Descrição: Escreva uma função que encontre a projeção de um vetor sobre outro vetor.

```python
def projecao_a_em_b(a,b):

    #Calcula o produto interno entre os vetores
    dot_p=sum([a[i] * b[i] for i in range(len(a))])

    #calculo da norma do vetor b
    norma_b=np.sqrt(sum([b[i]**2 for i in range(len(b))]))
    
    #Calcula o fator de escala da projeção de a em b
    fator_escala = dot_p/norma_b

    #Calcula o vetor  projeção de a em b
    return [fator_escala*i for i in b]
```
