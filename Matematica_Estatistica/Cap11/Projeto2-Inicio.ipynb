## Parte 1 - Escrever o Algoritmo

```python
#Classe para definição do algoritimo
class AlgoritimoNeuralNetworkDSA:
    
    #Metodo construtor(Inicia atributos das classe quando o metodo for criado)
    def __init__(self,taxa_aprendizado,num_iteracoes):
        
        #inicia atributos da classe com os valores passados como argumento
        self.taxa_aprendizado=taxa_aprendizado
        self.num_iteracoes=num_iteracoes

        #inicia coeficientes com None (isto é o que o modelo ira aprender)
        self.pesos=None
        self.bias=None

    #Metodo da função de ativação sidmoide para gerar a predição no formato que pode ser interpretado
    #Como probabilidade (valores de 0 a 1 para cada classe prevista)
    def func_activation_sigmoid(self,pred):

        #Função Sigmoide
        sig = 1 / ( 1 + np.exp(-pred))
        return sig

    #Metodo de treinamento
    def fit(self,X,y):
        
        #Extrai o numero de linhas e colunas dos dados:
        num_registros, num_atributos= X.shape

        #Inicia a Matriz de pesos com zeros, com o mesmo shape do número de atributos 
        self.pesos = np.zeros(num_atributos)
        
        #Inicia o escalar bias com o valor zero
        self.bias=0

        print("\nTreinamento Iniciado!")

        #Loop de treinamento pelo numero de itarações
        for i in range(self.num_iteracoes):

            #print("Iteração: ",i)

            #Primeira parte do metodo fit é o Forward Pass
            
            #Faz a previsão usando os dados de entrada, pesos e bias
            previsao = np.dot(X,self.pesos) +self.bias
            #print("Previsão antes da função de ativação: ",previsao)

            #Converte a previsão com a função de ativação para um valor que pode ser interpretado
            previsao_final = self.func_activation_sigmoid(previsao)
            #print("Previsão depois da função de ativação: ",previsao_final)

            #Calcula o erro do modelo
            erro = (previsao_final - y)
            #print("Erro do modelo: ",erro)

            #Segunda parte do metodo fit é a passada para trás, Back Propagation

            #Calcula os Gradientes (derivadas de pesos e de bias)
            dw = (1 / num_registros) * np.dot(X.T,erro)
            db = (1 / num_registros) * np.sum(previsao_final - y)

            #Atualiza pesos e bias utilizando os Gradientes e a taxa de aprendizado

            #Formula: pesos = pesos- derivada de pesos * taxa de aprendizado
            self.pesos -= self.taxa_aprendizado * dw
            #print("Valores de Pesos: ",self.pesos)

            #Formula: bias= bias - derivada de bias * taxa de aprendizado
            self.bias -= self.taxa_aprendizado * db
            #print("Valores de Bias: ",self.bias)

        print("\nTreinamento Concluído!")

    #Metodo para realizar previsões
    def predict(self,X):

        #Faz a previsão com novos dados de entrada e com valores treinados de  pesos e bias
        previsao = np.dot(X,self.pesos) + self.bias
        #print("Previsão antes da função de ativação: ",previsao)

        #Converte a previsão no formato de probabilidades atraves da função sigmoide
        previsao_final = self.func_activation_sigmoid(previsao)
        #print("Previsão depois da função de ativação: ",previsao_final)

        #Aplica o "Cut-off" e converte probabilidades para classes binarias (0 ou 1)
        classe_prevista = [1 if i > 0.5 else 0 for i in previsao_final]

        return classe_prevista
             
```

## Parte 2 - Preparar os Dados de Treino e Teste

```python
taxa_aprendizado=0.01
num_iteracoes=1000
```

```python
modeloDSA=AlgoritimoNeuralNetworkDSA(taxa_aprendizado=taxa_aprendizado,num_iteracoes=num_iteracoes)
```

```python
#Os dados de entrada representam transações comerciais
#Temos 2 atributos e 8 registros
dados_entrada = np.array([[1, 2.5],[2, 3],[3, 5],[1, 4],[1.5, 2],[5, 6],[6, 7],[4, 5.5]])
```

## Parte 3 - Treinamento do Modelo

```python
# Dados de entrada e saída para treinar o modelo 
X_treino = np.array([[1, 2.5], [2, 3], [3, 5], [1, 4], [5, 6], [6, 7]])
y_treino = np.array([0, 0, 1, 0, 1, 1])
```

```python
modeloDSA.fit(X_treino,y_treino)
```

## Parte 4 - Avaliação do Modelo

```python
# Dados de teste (2 registros)
X_teste = np.array([[1.5, 2], [4, 5.5]])
y_teste = np.array([0, 1])
```

```python
X_teste
```

```python
previsoes_teste = modeloDSA.predict(X_teste)
```

```python
# Bloco if para avaliar os resultados 
for i, previsao in enumerate(previsoes_teste):
    
    # Extrai os valores reais de entrada e saída
    entrada = X_teste[i]
    saida = y_teste[i]
    
    # Checando o resultado
    if previsao == 0:
        print(f"\nEntrada {entrada} e saída {saida} a previsão foi {previsao}: Não é Transação Suspeita")
    else:
        print(f"\nEntrada {entrada} e saida {saida} a previsão foi {previsao}: É Transação Suspeita")

print("\n")
```
