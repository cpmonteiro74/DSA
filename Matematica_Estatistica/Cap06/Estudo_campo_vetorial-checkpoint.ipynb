```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Cria a figura
plt.figure(figsize = (5, 5))

# Cria um array de vetores
setas=12
x=np.cos(np.linspace(0, (2*np.pi), setas+1))
y=np.sin(np.linspace(0, (2*np.pi), setas+1))
fasor=np.c_[x,y]
print('Vetores:')
print(fasor)

# Cria um array de vetores para as origens
origens=np.c_[np.zeros((1,setas+1)),np.zeros((1,setas+1))].reshape(2,setas+1)
print('\nOrigens:')
print(origens)

# Plot
color=['r','g','b']
plt.quiver(*origens, fasor[:,0], fasor[:,1], color=int(setas/3)*color,scale = 2)

# Escala
plt.axis([-1, 1, -1, 1])
plt.grid()
#plt.xlim(-1,1)
#plt.ylim(-1,1)

# Salva o gráfico
plt.savefig('imagens/Estudo.png', dpi = 300) 

# Plot
plt.show()
```
