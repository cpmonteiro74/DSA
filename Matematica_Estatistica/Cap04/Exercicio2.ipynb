### Exercício 1: Gráfico de Uma Função Linear

Crie um gráfico para a função y = 2x + 3 para valores de x no intervalo de 5 a 10.

```python
import numpy as np
import matplotlib.pyplot as plt
import math
```

```python
def f(x):
    return 2*x+3

x=np.linspace(5,10,10)
y=f(x)

plt.figure(figsize = (6, 4))
plt.plot(x,y,label='f(x)=2x+3')
plt.grid()
plt.xlabel('x')
plt.ylabel('f(x)')
plt.legend()
plt.title('Grafico de uma função linear')
```

### Exercício 2: Gráfico de uma Função Quadrática

Desenhe o gráfico da função y = x^2 - 4x + 4. Crie uma massa de dados de exemplo usando np.linspace().

```python
def f(x):
    return x**2+2*x+3

x=np.linspace(-10,10,100)
y=f(x)

plt.figure(figsize = (6, 4))
plt.plot(x,y,label='f(x)=x²-4x+4')
plt.grid()
plt.xlabel('x')
plt.ylabel('f(x)')
plt.legend()
plt.title('Grafico de uma função quadratica')
```

### Exercício 3: Gráfico de Uma Função Senoidal

Crie um gráfico da função y = sin(x) para valores de x no intervalo de 0 a 2π.

```python
def f(x):
    return np.sin(x)

x=np.linspace(0,2*np.pi,100)
y=f(x)

plt.figure(figsize = (6, 4))
plt.plot(x,y,label='f(x)=sin(x)')
plt.grid()
plt.xlabel('x')
plt.ylabel('f(x)')
plt.legend()
plt.title('Gráfico de uma função senoidal')
```

### Exercício 6: Resolver uma Equação Quadrática

Escreva um script em Python que resolve a equação x^2 − 5x + 6 = 0 e imprime as raízes.

```python
import cmath
a=1
b=-5
c=6
delta=b**2-4*a*c
x1=(-b-cmath.sqrt(delta))/2*a
x2=(-b+cmath.sqrt(delta))/2*a
print(x1,x2)
```

### Exercício 10: Aplicação de Logaritmos

Utilize logaritmo para calcular o tempo necessário para que um investimento atinja um determinado valor, considerando uma taxa de crescimento contínua.

```python
#M=C*(1+i)^t
#M/C=(1+i)^t
#t=log(M/C,(1+i))
c=float(input("Capital Investido"))
i=float(input("Taxa de juros em %"))
m=float(input("Montante esperado"))
print('Para um investimento de {} atingir o montante {} a uma taxa {} m.m/a.a são necessários {} meses/anos'.format(c,m,i,int(math.log(m/c,(1+i/100)))))
```

```python
c=float(input('entre com o capital inicial'))
m=float(input('entre com o montante esperado'))
```

```python
i=[0.5,0.75,1,1.5]
plt.figure(figsize = (6, 4))

for i in i:    
    x=np.linspace(0,int(math.log(m/c,(1+i/100))),100)
    y=c*(1+i/100)**x
    plt.plot(y,x,label=str(i)+'%')
    plt.annotate(str(int(x[99])),(y[99],x[99]))

plt.grid()
plt.legend()
plt.title("Tempo de retorno pela taxa de juros")
```
