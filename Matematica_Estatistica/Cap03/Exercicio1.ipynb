### Exercício 2: Conversão de Temperatura

Crie uma função que converte a temperatura de Celsius para Fahrenheit e vice-versa. O usuário deve informar a temperatura e a escala de origem.

```python
#(0°C × 9/5) + 32 = 32°F
def temp_converter(escala,t):
    if escala=='C':
        print (t,' em Ceusis é: ',t*9/5 + 32,' F')
    elif escala=='F':
        print (t,' em Fahrenheit é: ',(t-32)*5/9,' C')
```

```python
temp_converter('F',80)
```

### Exercício 4: Cálculo de Média Aritmética

Escreva um programa que calcula a média aritmética de uma lista de números fornecida pelo usuário.

```python
def media_arit(lista=[]):
    e=0
    s=0
    for n in lista:
       e=e+1
       s=s+n
    print(s/e)
    
```

```python
media_arit([2,4])
```

### Exercício 5: Determinar as Raízes de uma Equação Quadrática

Crie um programa que encontra as raízes de uma equação quadrática do tipo ax^2 + bx + c = 0. O programa deve lidar com todos os tipos de raízes (reais e complexas).

```python
from numpy import complex_ as cpx
```

```python
def equad(a,b,c):
    d=b**2-4*a*c
    x1=(-b-cpx(pow(d,1/2)))/2*a
    x2=(-b+cpx(pow(d,1/2)))/2*a
    print (x1,x2)

print('Solução de Equações de 2 Grau: x^2 + bx + c = 0')
a=float(input('entre com valor de a: '))
b=float(input('entre com valor de b: '))
c=float(input('entre com valor de c: '))
print('As raizes x1 e x2 são respectivamente')
equad(a,b,c)
```

### Exercício 7: Sequência de Fibonacci

Crie um programa que gera a sequência de Fibonacci até um determinado número n fornecido pelo usuário.

```python
def fibo(n):
    l=[]
    for i in range(n):
        if i<2:
            l.append(1)
        else:
            l.append(l[i-1]+l[i-2])
    print(l)
```

```python
fibo(30)
```

### Exercício 8: Cálculo de Juros Compostos

Faça um programa que calcula o montante final de um investimento com juros compostos. O usuário deve fornecer o capital inicial, a taxa de juros anual, e o número de anos.

```python
def juros_comp(capital=1000,juros=1,anos=10):
    """
    Calcula o montante 'M' atraves da formula de juros compostos
    M=C*(1+j)**a, onde 
    'C' : capital inicial,
    'j' : juros em %
    'a' : tempo decorrido em anos 
    \n
    """
    m=capital*pow(1+juros/100,anos)
    print(m)
```

```python
juros_comp(10000,12,20)
```

### Exercício 10: Verificar se um Número é Primo

Crie uma função que verifica se um número fornecido pelo usuário é primo ou não.

```python
def numeros_primos(p):
    t=0
    for i in range(2,p):
        if p%i==0:
            t=1
            
    if t>0:
        print(p," Não é primo")
    else:
        print(p," é primo")
```

```python
numeros_primos(10)
```
