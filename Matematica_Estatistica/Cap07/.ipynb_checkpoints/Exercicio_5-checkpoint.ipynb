### Exercício 1: Cálculo de Distância Euclidiana

Descrição: Escreva uma função em Python que receba dois vetores (listas de números) de 
mesma dimensão e retorne a distância euclidiana entre eles.

```python
def distancia_euclidiana(a,b):
    return np.linalg.norm(np.array(a)-np.array(b))
```

```python
a=[1,2,4]
b=[1,2,4]
```

```python
distancia_euclidiana(a,b)
```

### Exercício 2: Distância Euclidiana com Validação de Dados

Descrição: Modifique a função do Exercício 1 para validar se os dois vetores têm o 
mesmo tamanho. Se não tiverem, a função deve retornar um erro.

```python
def distancia_euclidiana(a,b):
    if len(a)==len(b):
        return np.linalg.norm(np.array(a)-np.array(b))
    else:
        print('Erro, vetores com dimensões diferentes')
```

```python
distancia_euclidiana(a,b)
```

```python
distancia_euclidiana([1,2],[3,4,4])
```

### Exercício 3: Distância Euclidiana em N-Dimensões

Descrição:  Generalize  a  função  do  Exercício  1  para  que  ela  possa  calcular  a  distância 
euclidiana entre dois vetores em um espaço de qualquer número de dimensões.

```python
def distancia_euclidiana(a,b):
    if len(a)==len(b):
        return np.linalg.norm(np.array(a)-np.array(b))
    else:
        print('Erro, vetores com dimensões diferentes')
```

### Exercício 4: Comparando Distâncias

Descrição:  Escreva  um  script  que  receba  três  vetores  de  mesma  dimensão.  Calcule  a 
distância  euclidiana  entre  o  primeiro  e  os  outros  dois  vetores.  Imprima  qual  dos  dois 
vetores está mais próximo do primeiro, baseado na distância euclidiana.

```python
vetores={'a':[1,2,4,3],'b':[1,3,8,7],'c':[3,4,5,6]}
```

```python
# Dicionários para calcular a distância euclidiana
euclid = {}
# Loop para calcular a norma dos vetores
for k in vetores.keys():
    
    # Target
    vecA = np.array(vetores['a'])
    
    # País
    vecB = np.array(vetores[k])
    
    # Calcula a distância euclidiana com a operação entre os vetores
    dist = np.linalg.norm(vecA - vecB)
     
    
    # Carrega os dicionários
    euclid[k] = dist    
    
```

```python
# Coloca os dicionários em um dataframe
df_distance = pd.DataFrame({"Distancia Euclidiana": euclid})
```

```python
print(df_distance["Distancia Euclidiana"].sort_values()[1:].head(1))
```

## Exercício 5: Aplicação em Data Science

Descrição: Dada uma matriz de  dados (lista de listas, onde cada sublista é um vetor de 
características)  e  um  vetor  de  referência,  escreva  uma  função  que  calcule  a  distância 
euclidiana de cada vetor na matriz até o vetor de referência. A função deve retorna r uma lista de distâncias.

```python
matriz_dados=[[1,2,4,5],[2,4,3,5],[9,4,3,2],[3,2,3,1]]
```

```python
vetor_ref=[1,3,5,4]
```

```python
def distancias(matriz_dados,vetor_ref):
    distancias=[]
    for v in matriz_dados:
        distancias.append(np.linalg.norm(np.array(vetor_ref)-np.array(v)))
    return distancias
```

```python
distancias(matriz_dados,vetor_ref)
```

```python
np.linalg.norm([1,1])
```

```python
np.linalg.norm([2,3])
```

```python
np.linalg.norm(np.array([1,1])-np.array([2,3]))
```
