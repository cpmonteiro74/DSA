```python
# Imports
import scipy
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import uniform
from scipy.stats import expon
from scipy.stats import norm
from scipy.integrate import quad
```

### Exercício 1: Tempo de Produção

Suponha  que  o  tempo  de  produção  de  um  carro  em  uma  fábrica  siga  uma  distribuição 
normal  com  uma  média  de  20  horas  e  um  desvio  padrão  de  2  horas.  Pergunta:  Qual  é  a 
probabilidade de que um carro seja 
produzido em menos de 18 horas?

```python
mu , sigma = 20,2
x = np.linspace(mu - 3 * sigma, mu + 3 * sigma, 100)
pp=18

p18 = norm.cdf(pp, mu, sigma)

# Plotando a função de densidade de probabilidade
plt.plot(x, norm.pdf(x, mu, sigma), label = 'PDF')

# Preenchendo a área sob a curva no intervalo [a, b]
px = np.linspace(0, pp, 100)
py = norm.pdf(px, mu, sigma)
plt.fill_between(px, py, color = 'grey', alpha = 0.5)

# Adicionando anotações no gráfico
plt.text(20, 0.025, f'Área = {p18:.2f}', horizontalalignment = 'center')
plt.title('Distribuição Normal e Probabilidade < 18 horas')
plt.xlabel('Valores')
plt.ylabel('Probabilidade')
plt.legend()
plt.xlim(10,30)
plt.grid(True)
plt.show()

print('\nA probabilidade de que um carro seja produzido em menos de 18 horas é: ',p18*100,'%')

```

### Exercício 2: Consumo de Combustível

O   consumo   de   combustível   (em   km/l)   de   um   determinado   modelo   de   carro   é 
normalmente distribuído com uma média de 12 km/l e um desvio padrão de 3 km/l. 
Pergunta: Qual  é  a  probabilidade  de  que  um  carro  escolhido  aleatoriamente  tenha  um  consumo  de 
combustível maior que 13 km/l?

```python
mu , sigma = 12,3
x = np.linspace(mu - 3 * sigma, mu + 3 * sigma, 100)
pp=13

p13 =1- norm.cdf(pp, mu, sigma)

# Plotando a função de densidade de probabilidade
plt.plot(x, norm.pdf(x, mu, sigma), label = 'PDF')

# Preenchendo a área sob a curva no intervalo [a, b]
px = np.linspace(pp, 25, 100)
py = norm.pdf(px, mu, sigma)
plt.fill_between(px, py, color = 'grey', alpha = 0.5)

# Adicionando anotações no gráfico
plt.text(20, 0.025, f'Área = {p13:.2f}', horizontalalignment = 'center')
plt.title('Distribuição Normal e probabilidade >13 km/l')
plt.xlabel('Valores')
plt.ylabel('Probabilidade')
plt.legend()
#plt.xlim(3,27)
plt.grid(True)
plt.show()

print(f'\nA probabilidade de um consumo de combustível maior que 13 km/l é de: {p13*100:.2f}%')
```

### Exercício 3: Peso dos Automóveis

O peso dos automóveis produzidos por uma fábrica segue uma distribuição normal com 
uma média de 1500 kg e um desvio padrão de 50 kg. Pergunta: Qual é a probabilidade de que 
um automóvel produzido tenha um peso entre 1450 kg e 1550 kg?

```python
mu , sigma = 1500,50
x = np.linspace(mu - 3 * sigma, mu + 3 * sigma, 100)
a,b=1450,1550

pab=norm.cdf(b,mu,sigma)-norm.cdf(a,mu,sigma)

# Plotando a função de densidade de probabilidade
plt.plot(x, norm.pdf(x, mu, sigma), label = 'PDF')

# Preenchendo a área sob a curva no intervalo [a, b]
px = np.linspace(a, b, 100)
py = norm.pdf(px, mu, sigma)
plt.fill_between(px, py, color = 'grey', alpha = 0.5)

# Adicionando anotações no gráfico
plt.text(1500, 0.005, f'Área = {pab:.2f}', horizontalalignment = 'center')
plt.title('Distribuição Normal e Área sob a Curva')
plt.xlabel('Valores')
plt.ylabel('Probabilidade')
plt.legend()
#plt.xlim(3,27)
plt.grid(True)
plt.show()

print(f'\na probabilidade de que um automóvel produzido tenha um peso entre 1450 kg e 1550 kg é: {pab*100:.2f}%')
```

## Exemplo 4: Nota Média na Prova
